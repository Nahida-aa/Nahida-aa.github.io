### 波次1 2012

````python
import pandas as pd

def get_data(path, vars_dict):
    """
    读取数据
    :param path: 数据路径
    :param vars_dict: 变量字典 {原变量名: 新变量名} (采用扩展字段的方式)
    :return: 数据
    """
    data = pd.read_stata(path, convert_categoricals=False)
    # 打印数据的基本信息
    print(data.info())
    print(data.columns)
    # 选择需要的变量
    data = data[list(vars_dict.keys())]
    # 重命名变量
    data.rename(columns=vars_dict, inplace=True)
    return data

rename1id = lambda x: x[:9] + '0' + x[9:] if len(x) > 9 else x
one_otherTo1_0 = lambda x: 1 if '1' in str(x) else 0
one_otherTo0_1 = lambda x: 0 if '1' in str(x) else 1 
# 1~4, 5~10 -> 0, 1
one_4_otherTo0_1 = lambda x: 0 if x in [1.0, 2.0, 3.0, 4.0] else 1

demographic1_path = r"C:\Users\aa\Downloads\charls_data\demographic_background_1\demographic_background.dta" # 2011-2012 基本信息
# 波次 1
# 提取并重命名第1个数据集的列
columns={'ID': 'id',
    'ba002_1': 'birthyear_1', # 出生年份
    'rgender': 'gender_1', # 1:男, 2:女 -> male, female
    'be001': 'marital_1', # 1:已婚, ... -> 0:已婚(Married), 1:Others
    'bd001': 'edu_1', # 值为 1 到 5（未完成小学、小学、初中）, 值为 6 到 10（高中及以上） -> 0, 1
    'bc001': 'hukou_1', # 户口: 1:农村, other -> 0:农村(Rural), 1:其他(Other)
}
demographic_df_1 = get_data(demographic1_path, columns)

# 实现 1:男, 2:女 -> male, female
demographic_df_1['gender_1'] = demographic_df_1['gender_1'].apply(lambda x: 'Male' if x == 1 else 'Female')
demographic_df_1['marital_1'] = demographic_df_1['marital_1'].apply(lambda x: 'Married' if x == 1 else 'Others') # 0:已婚(Married), 1:Others
demographic_df_1['edu_1'] = demographic_df_1['edu_1'].apply(one_4_otherTo0_1) # 0:未完成小学、小学、初中, 1:高中及以上
demographic_df_1['edu_1'] = demographic_df_1['edu_1'].apply(lambda x: 'Lower_level' if x == 0 else 'Higher_level') # 0:未完成小学、小学、初中, 1:高中及以上 -> Lower level, Higher level
demographic_df_1['hukou_1'] = demographic_df_1['hukou_1'].apply(lambda x: 'Rural' if x == 1 else 'Others') # Rural, Others
# 计算 2012 年和 2015 年的年龄
demographic_df_1['age_1'] = 2012 - demographic_df_1['birthyear_1']
# demographic_df_1['age_3'] = 2015 - demographic_df_1['birthyear_1']

print(demographic_df_1.info())
# print(demographic_df_1.head())
print(demographic_df_1)

# 计算 1_edu 的比例
edu_counts = demographic_df_1['edu_1'].value_counts(normalize=True)
print(edu_counts)
````

````python
health1_path = r"C:\Users\aa\Downloads\charls_data\health_status_and_functioning1\health_status_and_functioning.dta"
columns={
    'ID': 'id',
    'da007_7_': 'heart_1', # 1:yes, 2:no
    'da009_7_': 'heart_time_type_1', # 1:year, 2:age
    'da009_1_7_': 'heart_year_1',
    'da009_2_7_': 'heart_age_1',
    'da007_8_': 'stroke_1', # 1:yes, 2:no
    'da009_8_': 'stroke_time_type_1', # 1:year, 2:age
    'da009_1_8_': 'stroke_year_1',
    'da009_2_8_': 'stroke_age_1',
    'da007_2_': 'dyslipidemia_1', # 血脂异常 1:yes, 2:no
    'da009_2_': 'dyslipidemia_time_type_1', # 1:year, 2:age
    'da009_1_2_': 'dyslipidemia_year_1',
    'da009_2_2_': 'dyslipidemia_age_1',
    'da007_1_': 'hypertension_1', #高血压 1:yes, 2:no  
    'da009_1_': 'hypertension_time_type_1', # 1:year, 2:age
    'da009_1_1_': 'hypertension_year_1',
    'da009_2_1_': 'hypertension_age_1',
    'da007_3_': 'diabetes_1', #糖尿病 1:yes, 2:no
    'da007_4_': 'cancer_1', #癌症 1:yes, 2:no
    'da059': 'smoke_1', # 1:吸烟, 2:不吸烟
    'da061': 'smoking_1', # 1:仍在吸烟, 2:已戒烟, None:不吸烟
    'da069': 'drink_1', # 1:从不, 2\3: others
}
health1_df = get_data(health1_path, columns)

columns_to_apply = ['heart_1', 'stroke_1', 'dyslipidemia_1', 'hypertension_1', 'diabetes_1', 'cancer_1']  # 需要应用函数的列名
for col in columns_to_apply:
    health1_df[col] = health1_df[col].apply(one_otherTo1_0)
# 新增字段 1_CVD = 1_heart or 1_stroke
health1_df['CVD_1'] = health1_df.apply(
    lambda row: 1 if (row['heart_1'] == 1 or row['stroke_1'] == 1) else 0, axis=1
)
# 计算 高血压(hypertension) 的比例
hypertension_counts_1 = health1_df['hypertension_1'].value_counts(normalize=True)
print(hypertension_counts_1)
# 计算 血脂异常(dyslipidemia) 的比例
dyslipidemia_counts_1 = health1_df['dyslipidemia_1'].value_counts(normalize=True)
print(dyslipidemia_counts_1)

# 将吸烟状态分为三类： 1_smoking_status: 从未吸烟、以前吸烟、当前吸烟
# 定义分类函数
def classify_smoking(row):
    if row['smoke_1'] == 2:
        return 'Never'
    elif row['smoke_1'] == 1 and row['smoking_1'] == 2:
        return 'Former'
    elif row['smoke_1'] == 1 and row['smoking_1'] == 1:
        return 'Current'
    # else: 不处理就是相当于 None
    #     return 'Unknown'  # 处理缺失值或异常情况
health1_df['smoking_status_1'] = health1_df.apply(classify_smoking, axis=1)
# 计算 1_smoking_status 的比例
smoking_status_1 = health1_df['smoking_status_1'].value_counts(normalize=True)
print(smoking_status_1)

health1_df['drinking_status_1'] = health1_df['drink_1'].apply(lambda x: 'Others' if x == 1 else 'Current') # 1:从不, 2\3: others -> Current, Others
# print(health1_df.info())
print(health1_df)
````

## PHR,中风,CVD死亡率的关系：一项基于人群的观察性研究
2024 年 4 月 24

方法:
1. 在横断面调查期间获得卒中史、 HDL-C 水平和血小板计数
2. PHR 计算为血小板数量与 HDL-C 浓度的比率, 即 PC （1000 个细胞/μL）/ HDL-C （mmol/L）
3. 加权 logistic 回归用于评估 HDL-C 和 PHR 与卒中的相关性
4. 这种关系的非线性是通过限制三次样条（RCS）和两段线性回归来确定的，用于识别拐点
5. 采用 Cox 回归前瞻性分析 PHR 和 HDL-C 浓度与中风幸存者心血管疾病 （CVD） 死亡率的相关性

结果:
1. 完全调整后，与 PHR 每标准差 （SD） 增加相关的卒中比值比 （OR） 估计为 1.13 （95% 置信区间 （CI）：1.03 − 1.24，P = 0.01）， 与 HDL-C 每 SD 增加相关的卒中 OR 为 0.95 （95% CI： 0.86-1.05，P = 0.30）
2. RCS 表明两个变量呈非线性关系 （P_PHR = 0.018 和 P_HDL-C = 0.003），进一步的分段线性回归确定了 PHR = 223.684 和 HDL-C = 1.4 mmol/L 的拐点, 节段回归表明，在 PHR ≥ 223.684 段中，与 PHR 每 SD 增加相关的卒中估计 OR 为 1.20 （95% CI： 1.09 − 1.31，P < 0.001），而拐点前后卒中与 HDL-C 的相关性不显著 （P > 0.05）
3. Cox 回归和 RCS 显示，PHR 的每 SD 增加与卒中幸存者 CVD 死亡风险增加呈线性相关 （HR： 1.14,95% CI： 1.06 − 1.22，P < 0.001;非线性， P = 0.956），而 HDL-C 与 CVD 死亡率无显著相关性

结论:
PHR 与卒中发生率之间的关联表现出显着的阈值效应，拐点为 223.684。PHR 超过 223.684 与中风呈正相关，而 HDL-C 与中风之间的关联不显著。此外，PHR 与卒中幸存者的 CVD 死亡率呈正线性相关

#### RSF
RSF（Random Survival Forest） 和 Cox 回归模型（Cox Proportional Hazards Model） 都是用于生存分析的模型，但它们的理论基础、适用场景和实现方式有显著的区别

````python
import pandas as pd
from sksurv.util import Surv
from sksurv.ensemble import RandomSurvivalForest
from sklearn.model_selection import train_test_split
from sksurv.metrics import concordance_index_censored
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# 加载数据
# df = pd.read_csv("charls.csv", usecols=fields_to_keep)

# --------------------------
# 第一部分：波次1数据处理
# --------------------------
# 筛选波次1未患CVD的样本
df_wave1 = df[df["CVD_1"] == 0].copy()

# 计算生存时间与事件状态
cvd_cols_post1 = ["CVD_2", "CVD_3", "CVD_4", "CVD_4"]
wave_years_post1 = [2013, 2015, 2018, 2020]

def calculate_survival(row):
    for i, col in enumerate(cvd_cols_post1):
        if row[col] == 1:
            return (wave_years_post1[i] - 2012, 1)  # (时间, 事件)
    return (2020 - 2012, 0)  # 删失数据

df_wave1[["survival_time", "cvd_status"]] = df_wave1.apply(
    lambda row: calculate_survival(row), axis=1, result_type="expand"
)

# 计算PHR
df_wave1["PHR_1"] = df_wave1["plt_1"] / df_wave1["hdl_1"]

# --------------------------
# 第二部分：分类变量处理
# --------------------------
# 定义特征与变量类型
features_wave1 = [
    # "age_1", 
    # "gender_1",          # 分类变量（二分类）
    # "marital_1",         # 分类变量（二分类）
    # "hukou_1",           # 分类变量（二分类）
    # "edu_1",             # 分类变量 (二分类)
    "smoking_status_1",  # 分类变量（多分类）
    "drinking_status_1", # 分类变量（二分类）
    # "hypertension_1",  # 高血压
    # "diabetes_1",  # 糖尿病
    # "dyslipidemia_1",  # 血脂异常
    # "cancer_1", # 癌症
    # "SBP_1", 
    # "DBP_1", 
    "tc_1", "tg_1",
    "ldl_1", "hdl_1", 
    "hba1c_1",
    "BMI_1",  "waist_1", 
    "PHR_1", 'PHR_diff', 'PHR_3', 'PHR_mean'
]

# 定义预处理流水线
# cat_features = ["gender_1", "marital_1", "hukou_1", "smoking_status_1", "drinking_status_1", "edu_1"]
cat_features = ["smoking_status_1", "drinking_status_1"]
# ordinal_features = []  # 有序变量需要单独处理

preprocessor = ColumnTransformer(
    transformers=[
        ("onehot", OneHotEncoder(drop="if_binary", handle_unknown="ignore"), cat_features),
        # ("ordinal", OrdinalEncoder(categories=[[1, 2, 3]]), ordinal_features),  # 明确指定教育等级顺序
    ],
    remainder="passthrough"  # 保留数值型变量
)

# --------------------------
# 第三部分：建模与评估
# --------------------------
# 准备数据
df_waveclean = df_wave1[features_wave1 + ["survival_time", "cvd_status"]].dropna()
X = df_waveclean[features_wave1]
y = Surv.from_dataframe("cvd_status", "survival_time", df_waveclean)

# 划分数据集
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y["cvd_status"], random_state=42
)

# 构建完整流水线（预处理 + 模型）
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomSurvivalForest(
        n_estimators=200,
        max_depth=5,
        min_samples_split=10,
        min_samples_leaf=5,
        random_state=42
    ))
])

# 训练模型
pipeline.fit(X_train, y_train)

# 评估性能
pred_risk = pipeline.predict(X_test)
c_index = concordance_index_censored(
    y_test["cvd_status"], y_test["survival_time"], pred_risk
)[0]
print(f"测试集C-index: {c_index:.3f}")

# --------------------------
# 第四部分：高级分析
# --------------------------
# 获取预处理后的特征名称
onehot_names = pipeline.named_steps["preprocessor"].named_transformers_["onehot"].get_feature_names_out(cat_features)
feature_names = np.concatenate([
    onehot_names,
    [f for f in features_wave1 if f not in cat_features]
])

from sklearn.inspection import permutation_importance
# 计算置换重要性
result = permutation_importance(
    pipeline.named_steps["model"],  # 模型
    pipeline.named_steps["preprocessor"].transform(X_test),  # 预处理后的测试集
    y_test,  # 测试集标签
    n_repeats=10,  # 重复次数
    random_state=42
)

# 构建重要性数据框
importance = pd.DataFrame({
    "feature": feature_names,
    "importance": result.importances_mean
}).sort_values("importance", ascending=False)

print("\n变量重要性排序:")
print(importance.head(10))
````

##### SHAP

SHAP 值本身不直接要求输入 survival_time 和 cvd_status，但需要基于已训练的生存模型（如RSF、Cox模型）的预测结果

KernelExplainer 是一种模型无关的 SHAP 方法，可以用于任何模型，但计算成本较高，尤其是当特征数量较多时

````python
import shap

# 提取预处理后的测试集数据
X_processed = pipeline.named_steps["preprocessor"].transform(X_test)

# 创建 SHAP KernelExplainer
explainer = shap.KernelExplainer(pipeline.named_steps["model"].predict, X_processed[:100])  # 使用部分数据作为背景数据

# 计算 SHAP 值
shap_values = explainer.shap_values(X_processed[:10])  # 仅计算前 10 个样本的 SHAP 值

# 绘制 SHAP 总结图
shap.summary_plot(shap_values, X_processed[:10], feature_names=feature_names)
````

1. 纵轴：按特征重要性从高到低排列的特征名称
2. 横轴：SHAP值的范围，表示特征对预测的影响方向（左为负向，右为正向）
3. 点的颜色：
  - 红色：该样本中该特征的实际值较高（例如年龄较大、PHR较高）。
  - 蓝色：该样本中该特征的实际值较低
4. 点的水平位置：
  - 右侧（SHAP值>0）：该特征增加了预测值（如CVD风险）。
  - 左侧（SHAP值<0）：该特征降低了预测值
- 关键观察点
  - 方向一致性：
    - 若某特征的红点集中在右侧，蓝点集中在左侧，说明高值特征与高风险正相关（如PHR越高，CVD风险越高）。
    - 若红点分布在左右两侧，可能表示非线性关系或阈值效应。

````python
# 将SHAP值转换为DataFrame（每行代表一个样本，每列代表一个特征）
shap_df = pd.DataFrame(shap_values, columns=feature_names, index=X_test.index[:10])

# 添加每个样本的预测基值（模型输出的平均值）和最终预测值
base_value = explainer.expected_value  # 基值
final_prediction = pipeline.named_steps["model"].predict(X_processed[:10])  # 最终预测值
shap_df["Base_Value"] = base_value
shap_df["Final_Prediction"] = final_prediction

# 打印前5个样本的SHAP值和预测结果
print("SHAP Values Table (Top 5 Samples):")
print(shap_df.head())

# 计算特征重要性（平均绝对SHAP值）
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": np.abs(shap_values).mean(axis=0)
}).sort_values("Importance", ascending=False)

print("\nFeature Importance Ranking:")
print(feature_importance)
````

````python
import statsmodels.api as sm
from patsy import dmatrix

# (df, x: {'label': 2012_PHR, 'value': 1_PHR}, y: {'label': CVD_2013_2020, 'value': CVD_2013_2020}) => None
def plot_RCS(df, x, y):
    knots = np.quantile(df[x['value']], [0.1, 0.5, 0.9]) 
    X = dmatrix("cr(x, knots=knots)", {"x": df[x['value']]}, return_type="dataframe")
    y = df[y['value']]

    # 拟合Logistic回归模型
    model = sm.Logit(y, X).fit()
    print(model.summary())
    # 可视化RCS曲线
    # 生成均匀覆盖PHR范围的序列（100个点）
    phr_range = np.linspace(df[x['value']].min(), df[x['value']].max(), 100)
    # 生成样条基函数（与模型训练时相同参数）
    X_smooth = dmatrix("cr(x, knots=knots)", {"x": phr_range}, return_type="dataframe")
    # 预测对应的CVD风险
    pred_risk = model.predict(X_smooth)
    # 绘制平滑RCS曲线
    plt.plot(phr_range, pred_risk, color="blue", linewidth=2, label="RCS Curve")
    # plt.xlabel(x['label'])
    # plt.ylabel(f"Predicted {y['label']} CVD Risk")
    # plt.title(f"RCS Plot for {x['label']}")
    # plt.title("Restricted Cubic Spline (RCS) Plot for PHR Threshold Detection")
    threshold = 14.2  # 假设通过模型确定的最佳阈值
    # 标记阈值点（示例值，需根据实际结果调整）
    plt.axvline(x=threshold, color="red", linestyle="--", label=f"Threshold={threshold}")
    # 添加数据点（可选，观察原始分布）
    plt.scatter(df[x['value']], model.predict(X), alpha=0.3, color="grey", label="Observed Data")

    plt.legend()
    plt.show()

plot_RCS(df, {'label': 'PHR (Baseline)', 'value': '1_PHR'}, {'label': 'Predicted 2013-2020 CVD Risk', 'value': 'CVD_2013_2020'})
plot_RCS(df, {'label': 'PHR (Follow-up)', 'value': '3_PHR'}, {'label': 'Predicted 2018-2020 CVD Risk', 'value': 'CVD_2018_2020'})
plot_RCS(df, {'label': 'PHR_mean', 'value': 'PHR_mean'}, {'label': 'Predicted 2015-2020 CVD Risk', 'value': 'CVD_2015_2020'})
````

````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from patsy import dmatrix
import matplotlib.pyplot as plt

def plot_RCS_with_CI(df, x_var, y_var, knots_quantiles=[0.1, 0.5, 0.9]):
    # 设置节点
    knots = np.quantile(df[x_var].dropna(), knots_quantiles)
    
    # 构建样条基函数
    X = dmatrix("cr(x, knots=knots)", {"x": df[x_var], "knots": knots}, return_type="dataframe")
    y = df[y_var]
    
    # 拟合Logistic回归（含协变量）
    formula = f"{y_var} ~ cr({x_var}, knots=knots) + age_1 + hypertension_1"
    model = sm.Logit.from_formula(formula, data=df).fit()
    print(model.summary())
    
    # 生成均匀PHR值用于预测
    phr_range = np.linspace(df[x_var].min(), df[x_var].max(), 100)
    X_smooth = dmatrix("cr(x, knots=knots)", {"x": phr_range, "knots": knots}, return_type="dataframe")
    
    # 预测概率及置信区间
    pred = model.get_prediction(X_smooth)
    pred_mean = pred.predicted_mean
    pred_ci = pred.conf_int(alpha=0.05)
    
    # 绘图
    plt.figure(figsize=(10,6))
    plt.fill_between(phr_range, pred_ci[:,0], pred_ci[:,1], color="blue", alpha=0.2, label="95% CI")
    plt.plot(phr_range, pred_mean, color="blue", linewidth=2, label="Predicted Risk")
    plt.xlabel(x_var)
    plt.ylabel("Probability of CVD")
    plt.title(f"RCS Plot for {x_var} with 95% CI")
    plt.legend()
    plt.show()

# 使用调整后的函数
plot_RCS_with_CI(df, "PHR_1", "CVD_2013_2020")
````

````python
# 检查是否有缺失值
# print(df[features].isnull().sum())
# 3645
# 检查数据分布
import matplotlib.pyplot as plt # pip install matplotlib
import seaborn as sns
from sklearn.cluster import KMeans # pip install seaborn
import numpy as np

def distribution(df, fields):
    # 使用直方图或核密度图查看数据分布是否存在异常值或偏态
    sns.histplot(df[fields[0]], kde=True, bins=30, color='blue', label=fields[0])
    sns.histplot(df[fields[1]], kde=True, bins=30, color='orange', label=fields[1])
    plt.legend()
    plt.show()

distribution(df, features)

# 使用箱线图查看异常值
for field in features:
    sns.boxplot(x=df[field])
    plt.title(f'Boxplot of {field}')
    plt.show()
# 使用 Z-Score 去除异常值
from scipy.stats import zscore
# 指定需要去除异常值的字段
# fields_to_check = ['1_BMI', '1_waist']
fields_to_check = features
z_scores = np.abs(zscore(df[fields_to_check]))
df_cleaned = df[(z_scores < 3).all(axis=1)].copy()  # 去除 Z-Score 大于 3 的异常值
print('去除 去除 Z-Score 大于 3 的异常值 后的数据分布:')
print(df_cleaned.shape)
# 查看 age_1 的 描述性统计
print(df_cleaned['age_1'].describe())
# 性别 计数 顺便 显示 百分比
print(df_cleaned['gender_1'].value_counts())
print(df_cleaned['gender_1'].value_counts(normalize=True))
# PHR
print(df_cleaned['PHR_1'].describe())
print(df_cleaned['PHR_3'].describe())
print(df_cleaned['cumulative_PHR'].describe())
# df_cleaned = df[
#     (df['1_BMI'] >= 10) & (df['1_BMI'] <= 50) &  # 假设 BMI 的合理范围是 10 到 50
#     (df['1_waist'] >= 40) & (df['1_waist'] <= 150)  # 假设腰围的合理范围是 40 到 150 cm
# ].copy()

distribution(df_cleaned, features)

# k-means 聚类 需要标准化数据 
from sklearn.preprocessing import StandardScaler # pip install scikit-learn
# 创建新的字段名
scaled_fields = [f"{field}_scaled" for field in features]
# 标准化数据并存储到新的字段
scaler = StandardScaler()
df_cleaned[scaled_fields] = scaler.fit_transform(df_cleaned[features])
# 检查结果
print(df_cleaned.head())

# 保存标准化后的数据
scaled_file_path = r"./scaled_data.csv"
df_cleaned.to_csv(scaled_file_path, index=False)

# 检查标准化后的数据分布
print('标准化后的数据分布:')
distribution(df_cleaned, scaled_fields)

# 相关性检查, 如果两个变量之间的相关性很高，可能会导致多重共线性问题
correlation = df_cleaned[features].corr()
print(correlation)

# 绘制肘部图，选择最佳的聚类数
from sklearn.metrics import silhouette_score
import numpy as np

# 绘制肘部图并计算轮廓系数
def plot_elbow_and_silhouette(data, fields, max_clusters=10):
    distortions = []
    silhouette_scores = []
    cluster_range = range(2, max_clusters + 1)  # 轮廓系数从 k=2 开始计算

    for k in cluster_range:
        kmeans = KMeans(n_clusters=k, random_state=42)
        kmeans.fit(data[fields])
        distortions.append(kmeans.inertia_)
        silhouette_scores.append(silhouette_score(data[fields], kmeans.labels_))

    # 绘制肘部图
    plt.figure(figsize=(14, 6))

    # 子图 1：肘部图
    plt.subplot(1, 2, 1)
    plt.plot(range(1, max_clusters + 1), [KMeans(n_clusters=k, random_state=42).fit(data[fields]).inertia_ for k in range(1, max_clusters + 1)], marker='o')
    plt.xlabel('Number of Clusters (k)')  # 聚类数
    plt.ylabel('Sum of Squared Errors (SSE)')  # 误差平方和
    plt.title('Elbow Method for Optimal k')  # 肘部法则

    # 子图 2：轮廓系数
    plt.subplot(1, 2, 2)
    plt.plot(cluster_range, silhouette_scores, marker='o', color='orange')
    plt.xlabel('Number of Clusters (k)')  # 聚类数
    plt.ylabel('Silhouette Score')  # 轮廓系数
    plt.title('Silhouette Score for Optimal k')  # 轮廓系数

    plt.tight_layout()
    plt.show()
def plot_elbow_and_silhouette2(data, fields, max_clusters=10):
    distortions = []
    silhouette_scores = []
    cluster_range = range(2, max_clusters + 1)  # 轮廓系数从 k=2 开始计算

    for k in cluster_range:
        kmeans = KMeans(n_clusters=k, random_state=42)
        kmeans.fit(data[fields])
        distortions.append(kmeans.inertia_)
        silhouette_scores.append(silhouette_score(data[fields], kmeans.labels_))

    # 单独绘制肘部图
    plt.figure(figsize=(7, 5))
    plt.plot(range(1, max_clusters + 1), [KMeans(n_clusters=k, random_state=42).fit(data[fields]).inertia_ for k in range(1, max_clusters + 1)], marker='o')
    plt.xlabel('Number of Clusters (k)')
    plt.ylabel('Sum of Squared Errors (SSE)')
    plt.title('Elbow Method for Optimal k')
    plt.grid()
    plt.show()

    # 单独绘制轮廓系数图
    plt.figure(figsize=(7, 5))
    plt.plot(cluster_range, silhouette_scores, marker='o', color='orange')
    plt.xlabel('Number of Clusters (k)')
    plt.ylabel('Silhouette Score')
    plt.title('Silhouette Score for Optimal k')
    plt.grid()
    plt.show()
# 绘制肘部图和轮廓系数
plot_elbow_and_silhouette(df_cleaned, features)
plot_elbow_and_silhouette2(df_cleaned, features)
````

##### KMeans
KMeans 聚类需要**事先确定簇数 (k)**。这是 KMeans 的一个特点，也是它的局限之一。你需要在算法运行之前指定要分成多少个簇，算法会根据这个簇数将数据划分。

###### 解决簇数选择问题的方法：
1. **肘部法 (Elbow Method)**：
   - 通过绘制误差平方和 (SSE) 随簇数变化的曲线，观察拐点位置来选择合适的簇数。

2. **轮廓系数 (Silhouette Score)**：
   - 计算每个样本的轮廓系数，评估不同簇数下的聚类质量，选择得分最高的簇数。

3. **Gap Statistic**：
   - 比较实际数据的聚类结果与随机分布数据的聚类结果，选择差异最大的簇数。

4. **领域知识**：
   - 根据数据的实际背景和业务需求，直接指定合理的簇数。

如果不确定簇数，或者希望自动确定簇数，可以考虑使用其他聚类算法，例如：
- **DBSCAN**：基于密度的聚类，不需要指定簇数。
- **Gaussian Mixture Model (GMM)**：可以通过模型选择（如 BIC 或 AIC）确定最佳簇数。

````python
from sklearn.decomposition import PCA
from mpl_toolkits.mplot3d import Axes3D
k = 3
kmeans = KMeans(n_clusters=k, random_state=42)
df_cleaned['cluster_k3'] = kmeans.fit_predict(df_cleaned[scaled_fields]) + 1 # 聚类结果从 1 开始

# 定义统一的颜色映射
palette = sns.color_palette('Set2', n_colors=k)

# 聚类结果可视化
def visualize_clusters(data, fields, cluster_col, palette):
    # 假设 X 是聚类特征，clusters 是聚类结果
    pca = PCA(n_components=2)
    X_scaled = data[fields].values  # 选择需要聚类的特征
    X_pca = pca.fit_transform(X_scaled)  # X_scaled 是标准化后的特征
    clusters = data[cluster_col].values  # 聚类结果
    # 绘制聚类结果
    plt.figure(figsize=(8, 6))
    for cluster_id in np.unique(clusters):
        plt.scatter(
            X_pca[clusters == cluster_id, 0],
            X_pca[clusters == cluster_id, 1],
            label=f'Cluster {cluster_id}',
            color=palette[cluster_id - 1],
            s=50
        )
    # 手动创建分类图例
    plt.legend(title="Cluster Types", loc="best")
    # 添加 PCA 解释的方差比例
    explained_variance = pca.explained_variance_ratio_
    plt.xlabel(f'PCA Component 1 ({explained_variance[0]*100:.2f}%)')
    plt.ylabel(f'PCA Component 2 ({explained_variance[1]*100:.2f}%)')
    plt.title('Clustering Results (PCA)')
    plt.show()
def visualize_clusters_3d(data, fields, cluster_col, palette):
    # 假设 X 是聚类特征，clusters 是聚类结果
    pca = PCA(n_components=3)
    X_scaled = data[fields].values  # 选择需要聚类的特征
    X_pca = pca.fit_transform(X_scaled)  # X_scaled 是标准化后的特征
    clusters = data[cluster_col].values  # 聚类结果

    # 绘制 3D 聚类结果
    # fig = plt.figure(figsize=(10, 6))
    fig = plt.figure(figsize=(14, 8))
    ax = fig.add_subplot(111, projection='3d')
    for cluster_id in np.unique(clusters):
        ax.scatter(
            X_pca[clusters == cluster_id, 0],
            X_pca[clusters == cluster_id, 1],
            X_pca[clusters == cluster_id, 2],
            label=f'Cluster {cluster_id}',
            color=palette[cluster_id-1],
            s=50
        )
    # 手动创建分类图例
    ax.legend(title="Cluster Types", loc="best")
    # 添加 PCA 解释的方差比例
    explained_variance = pca.explained_variance_ratio_
    ax.set_xlabel(f'PCA Component 1 ({explained_variance[0]*100:.2f}%)')
    ax.set_ylabel(f'PCA Component 2 ({explained_variance[1]*100:.2f}%)')
    ax.set_zlabel(f'PCA Component 3 ({explained_variance[2]*100:.2f}%)')
    # fig.tight_layout()
    ax.set_box_aspect([1, 1, 1])  # 设置为正方体比例
    plt.subplots_adjust(left=0.05, right=0.95, bottom=0.05, top=0.95)
    # ax.view_init(elev=20, azim=135)
    ax.set_title('Clustering Results (3D PCA)')
    plt.show()

def plot_cluster_means(data, cluster_col, palette, fields=['PHR_1','PHR_3']):
    # fields = ['PHR_1_scaled','PHR_3_scaled']
    # fields = ['1_PHR','3_PHR']
    # 计算每个聚类的均值
    cluster_means = data.groupby(cluster_col)[fields].mean()
    
    # 绘制每个聚类的均值
    plt.figure(figsize=(10, 6))
    for cluster, color in zip(cluster_means.index, palette):
        plt.plot(fields, cluster_means.loc[cluster], marker='o', label=f'Cluster {cluster}', color=color)
    
    plt.title('Cluster Means for PHR') # 聚类均值
    plt.xlabel('Fields') # 字段
    plt.ylabel('Mean PHR') # 平均 PHR
    plt.legend(title='Cluster')
    plt.grid(True)
    plt.show()
    # diff = cluster_means[fields[1]] - cluster_means[fields[0]]
    # 均值的 差值
    cluster_means['diff'] = cluster_means[fields[1]] - cluster_means[fields[0]]
    # 然后按聚类分组计算差的均值
    diff_means = data.groupby(cluster_col)['PHR_diff'].mean()
    # 将差的均值添加到输出中
    cluster_means['diff_mean'] = diff_means
    print(cluster_means)

# 可视化聚类结果
visualize_clusters(df_cleaned, scaled_fields, 'cluster_k3', palette)
visualize_clusters_3d(df_cleaned, scaled_fields, 'cluster_k3', palette)
# 绘制聚类均值图
# plot_cluster_means(df_cleaned,'cluster_k3', palette)
plot_cluster_means(df_cleaned,'cluster_k3', palette, scaled_fields[0:2])

# 对每个类别 进行 描述性统计
def cluster_descriptive_stats(data, cluster_col, fields):
    # 对每个聚类进行描述性统计
    for cluster in data[cluster_col].unique():
        print(f"\nCluster {cluster} Descriptive Statistics:")
        print(data[data[cluster_col] == cluster][fields].describe())
        print("\n")
````

````python
from scipy.stats import chi2_contingency, f_oneway, kruskal
# 处理连续变量
def process_continuous_variable(df, feature, cluster_col='cluster_k4'):
    results = {}
    # 计算整体统计
    overall_mean = df[feature].mean()
    overall_std = df[feature].std()
    results['Overall'] = f'{overall_mean:.2f} ± {overall_std:.2f}'
    
    # 计算分组统计
    groups = []
    for cluster in sorted(df[cluster_col].unique()):
        cluster_data = df[df[cluster_col] == cluster][feature]
        mean = cluster_data.mean()
        std = cluster_data.std()
        results[f'Cluster {cluster}'] = f'{mean:.2f} ± {std:.2f}'
        groups.append(cluster_data)
    
    # 计算 P 值（单因素方差分析）
    _, p_value = f_oneway(*groups)
    results['P Value'] = f'{p_value:.3f}'
    return results

# 分类变量
def process_categorical_variable(df, feature, cluster_col='cluster_k4'):
    results = {}
    # 计算整体统计
    overall_counts = df[feature].value_counts()
    overall_percentages = (overall_counts / len(df) * 100).round(1)
    overall_stats = [f'{value}: {count} ({percentage}%)' for value, count, percentage in zip(overall_counts.index, overall_counts, overall_percentages)]
    results['Overall'] = ', '.join(overall_stats)
    
    # 计算分组统计
    contingency_table = pd.crosstab(df[cluster_col], df[feature])
    for cluster in sorted(df[cluster_col].unique()):
        cluster_counts = contingency_table.loc[cluster]
        cluster_percentages = (cluster_counts / cluster_counts.sum() * 100).round(1)
        cluster_stats = [f'{value}: {count} ({percentage}%)' for value, count, percentage in zip(cluster_counts.index, cluster_counts, cluster_percentages)]
        results[f'Cluster {cluster}'] = ', '.join(cluster_stats)
    
    # 计算 P 值（卡方检验）
    _, p_value, _, _ = chi2_contingency(contingency_table)
    results['P Value'] = f'{p_value:.3f}'
    return results

age_stats = process_continuous_variable(df_cleaned, 'age_1', 'cluster_k3')
print('age:', age_stats)
gender_stats = process_categorical_variable(df_cleaned, 'gender_1', 'cluster_k3')
print('gender:', gender_stats)
marital_stats = process_categorical_variable(df_cleaned, 'marital_1', 'cluster_k3')
print('marital:', marital_stats)
edu_stats = process_categorical_variable(df_cleaned, 'edu_1', 'cluster_k3')
print('edu:', edu_stats)
hukou_stats = process_categorical_variable(df_cleaned, 'hukou_1', 'cluster_k3')
print('hukou:', hukou_stats)
smoking_status_stats = process_categorical_variable(df_cleaned, 'smoking_status_1', 'cluster_k3')
print('smoking_status:', smoking_status_stats)
drinking_status_stats = process_categorical_variable(df_cleaned, 'drinking_status_1', 'cluster_k3')
print('drinking_status:', drinking_status_stats)
hypertension_stats = process_categorical_variable(df_cleaned, 'hypertension_1', 'cluster_k3')
print('hypertension:', hypertension_stats)
diabetes_stats = process_categorical_variable(df_cleaned, 'diabetes_1', 'cluster_k3')
print('diabetes:', diabetes_stats)
dyslipidemia_stats = process_categorical_variable(df_cleaned, 'dyslipidemia_1', 'cluster_k3')
print('dyslipidemia:', dyslipidemia_stats)
cancer_stats = process_categorical_variable(df_cleaned, 'cancer_1', 'cluster_k3')
print('cancer:', cancer_stats)
sbp_stats = process_continuous_variable(df_cleaned, 'SBP_1', 'cluster_k3')
print('SBP:', sbp_stats)
diastolic_bp_stats = process_continuous_variable(df_cleaned, 'DBP_1', 'cluster_k3')
print('DBP:', diastolic_bp_stats)
print('HDL-C:', process_continuous_variable(df_cleaned, 'hdl_1', 'cluster_k3'))
print('LDL-C:', process_continuous_variable(df_cleaned, 'ldl_1', 'cluster_k3'))
print('TC:', process_continuous_variable(df_cleaned, 'tc_1', 'cluster_k3'))
print('TG:', process_continuous_variable(df_cleaned, 'tg_1', 'cluster_k3'))
print('hba1c:', process_continuous_variable(df_cleaned, 'hba1c_1', 'cluster_k3'))
print('PHR2012:', process_continuous_variable(df_cleaned, 'PHR_1', 'cluster_k3'))
print('PHR2015:', process_continuous_variable(df_cleaned, 'PHR_3', 'cluster_k3'))
print('cumulative_PHR:', process_continuous_variable(df_cleaned, 'cumulative_PHR', 'cluster_k3'))
print('cvd:', process_categorical_variable(df_cleaned, 'CVD', 'cluster_k3'))
````

````python
import numpy as np
from patsy import dmatrix
import statsmodels.api as sm
from scipy.stats import chi2

# 自动选择最优节点数的 RCS 模型
def select_best_rcs_model(data, y, x, min_knots=3, max_knots=5):
    best_aic = float('inf')  # 初始化最小 AIC
    best_model = None        # 最优模型
    best_knots = None        # 最优节点位置
    nonlinearity_p_value = None  # 非线性检验的 P 值
    
    # 尝试不同的节点数
    for num_knots in range(min_knots, max_knots + 1):
        # 计算节点位置（等分位数）
        knots = [np.percentile(data[x], p) for p in np.linspace(0, 100, num_knots + 2)[1:-1]]
        
        # 创建 RCS 特征
        rcs_basis = dmatrix(
            f"bs(data['{x}'], knots={knots}, degree=3, include_intercept=False)",
            {"data": data},
            return_type='dataframe'
        )
        
        # 拟合非线性模型(RCS)
        nonlinear_model = sm.Logit(data[y], rcs_basis)
        nonlinear_result = nonlinear_model.fit(disp=False)  # 关闭拟合过程的输出
        
        # 构建线性模型（仅包含线性项）
        X_linear = sm.add_constant(data[[x]])  # 添加常数项
        linear_model = sm.Logit(data[y], X_linear)
        linear_result = linear_model.fit(disp=False)
        
        # 计算似然比检验的 P 值
        lr_stat = 2 * (nonlinear_result.llf - linear_result.llf)  # 似然比统计量
        df_diff = nonlinear_result.df_model - linear_result.df_model  # 自由度差
        p_value = chi2.sf(lr_stat, df_diff)  # P 值
        
        # 比较 AIC，选择最优模型
        if nonlinear_result.aic < best_aic:
            best_aic = nonlinear_result.aic
            best_model = nonlinear_result
            best_knots = knots
            nonlinearity_p_value = p_value  # 保存非线性检验的 P 值
    
    # 输出非线性检验结果
    print(f"Nonlinearity Test P-Value: {nonlinearity_p_value:.3f}")
    if nonlinearity_p_value < 0.05:
        print("Significant nonlinearity detected. Threshold effect likely exists.")
    else:
        print("No significant nonlinearity detected. Threshold effect unlikely.")
    
    # 返回最优模型、节点信息和非线性检验 P 值
    return best_model, best_knots, nonlinearity_p_value

def plot_best_rcs_model(data, y, x, best_model, best_knots):
    # 生成预测数据
    x_values = np.linspace(data[x].min(), data[x].max(), 100)
    rcs_values = dmatrix(
        f"bs(x_values, knots={best_knots}, degree=3, include_intercept=False)",
        {"x_values": x_values},
        return_type='dataframe'
    )
    
    # 预测概率和置信区间
    X_pred = sm.add_constant(rcs_values)
    predictions = best_model.get_prediction(X_pred).summary_frame()
    predicted_probs = predictions['predicted']
    conf_int_lower = predictions['ci_lower']
    conf_int_upper = predictions['ci_upper']
    
    # 绘制曲线和置信区间
    plt.figure(figsize=(10, 6))
    plt.plot(x_values, predicted_probs, label='RCS Curve', color='blue')
    plt.fill_between(
        x_values, conf_int_lower, conf_int_upper,
        color='blue', alpha=0.2, label='95% CI'
    )
    plt.scatter(data[x], data[y], alpha=0.5, label='Data Points', color='orange')
    plt.xlabel(x)
    plt.ylabel(f'Probability of {y}')
    plt.title(f'Restricted Cubic Spline (RCS) for {y} Prediction with Optimal Knots')
    plt.legend()
    plt.grid()
    plt.show()

# 示例调用
lower_threshold = 6
df_r = df[df['cumulative_PHR'] > lower_threshold]  # 去除左侧极端值
best_model, best_knots, nonlinearity_p_value = select_best_rcs_model(df_r, 'CVD', 'cumulative_PHR')
print("Best Knots:", best_knots)
print(best_model.summary())
# 绘制最优模型
plot_best_rcs_model(df_r, 'CVD', 'cumulative_PHR', best_model, best_knots)
````

````python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 寻找 阈值点 (拐点)
def find_threshold_point(df, feature, target):
    # 计算分组统计
    df_cleaned = df.dropna(subset=[feature, target])
    
    # 计算分位数
    quantiles = np.linspace(0, 1, 100)
    thresholds = np.quantile(df_cleaned[feature], quantiles)
    
    # 初始化最小均方误差
    min_mse = float('inf')
    best_threshold = None
    
    for threshold in thresholds:
        # 创建分段变量
        df['segment'] = np.where(df[feature] > threshold, df[feature] - threshold, 0)
        
        # 拟合线性回归模型
        X = df[['segment']]
        y = df[target]
        model = LinearRegression().fit(X, y)
        
        # 计算均方误差
        mse = mean_squared_error(y, model.predict(X))
        
        if mse < min_mse:
            min_mse = mse
            best_threshold = threshold
            
    return best_threshold

df_r = df[df['cumulative_PHR'] > 6]  # 去除左侧极端值
print("cumulative_PHR 的最佳阈值点:", find_threshold_point(df_r, 'cumulative_PHR', 'CVD'))

# 描述 cumulative_PHR
print("cumulative_PHR 描述统计:")
print(df_r['cumulative_PHR'].describe())

# 候选拐点
candidate_knots = np.linspace(df_r['cumulative_PHR'].min(), df_r['cumulative_PHR'].max(), 100)
# candidate_knots = np.linspace(20, df['cumulative_PHR'].max(), 100)

best_knot = None
best_mse = float('inf')

for knot in candidate_knots:
    # 创建分段变量
    df_r['cumulative_PHR_segment'] = np.where(df_r['cumulative_PHR'] > knot, df_r['cumulative_PHR'] - knot, 0)
    
    # 拟合分段线性回归
    X = df_r[['cumulative_PHR', 'cumulative_PHR_segment']]
    y = df_r['CVD']
    model = LinearRegression().fit(X, y)
    
    # 计算均方误差
    mse = mean_squared_error(y, model.predict(X))
    if mse < best_mse:
        best_mse = mse
        best_knot = knot

print(f"Best Knot: {best_knot}")
````

拐点之前（cumulative_PHR <= knot）：

模型的预测值为： 
$$ \hat{y} = \text{Intercept} + \text{Coef}_{\text{cP}} \cdot \text{cP} $$

这表示在拐点之前， cumulative_PHR 对 CVD 的影响由 Coef_cumulative_PHR 决定。
拐点之后（cumulative_PHR > knot）：

模型的预测值为： 
$$ \hat{y} = \text{Intercept} + \text{Coef}{\text{cumulativePHR}} \cdot \text{cumulativePHR} + \text{Coef}{\text{segment}} \cdot (\text{cumulativePHR} - \text{knot}) $$
这表示在拐点之后，cumulative_PHR 对 CVD 的影响由 Coef_cumulative_PHR + Coef_segment 决定。

````python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

def fit_piecewise_logistic_model(df, knot):
    # 创建分段变量
    df['cumulative_PHR_segment'] = np.where(df['cumulative_PHR'] > knot, df['cumulative_PHR'] - knot, 0)
    
    # 拟合分段逻辑回归
    X = sm.add_constant(df[['cumulative_PHR', 'cumulative_PHR_segment']])  # 添加常数项
    y = df['CVD']
    model = sm.Logit(y, X).fit()
    return model

def plot_piecewise_logistic_regression(df, knot, model):
    # 创建分段变量
    df['cumulative_PHR_segment'] = np.where(df['cumulative_PHR'] > knot, df['cumulative_PHR'] - knot, 0)
    
    # 按 cumulative_PHR 升序排序
    df_sorted = df.sort_values(by='cumulative_PHR')
    
    # 绘制原始数据点
    plt.figure(figsize=(10, 6))
    plt.scatter(df_sorted['cumulative_PHR'], df_sorted['CVD'], alpha=0.5, label='Data Points', color='gray')
    
    # 绘制分段逻辑回归拟合曲线
    X = sm.add_constant(df_sorted[['cumulative_PHR', 'cumulative_PHR_segment']])
    y_pred = model.predict(X)
    plt.plot(df_sorted['cumulative_PHR'], y_pred, color='blue', label='Piecewise Logistic Fit')
    
    # 标注最佳拐点
    plt.axvline(x=knot, color='red', linestyle='--', label=f'Best Knot at {knot:.2f}')
    
    # 图形设置
    plt.xlabel('cumulative_PHR')
    plt.ylabel('Probability of CVD')
    plt.title('Piecewise Logistic Regression with Best Knot')
    plt.legend()
    plt.grid()
    plt.show()

# 示例调用
best_knot = 11.4933
piecewise_model = fit_piecewise_logistic_model(df, best_knot)
print(piecewise_model.summary())  # 打印模型摘要
plot_piecewise_logistic_regression(df, best_knot, piecewise_model)
````

##### 根据阈值分类

````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt

def fit_logistic_model(df, x_col, y_col):
    """
    拟合逻辑回归模型并返回结果
    """
    X = sm.add_constant(df[[x_col]])  # 添加常数项
    y = df[y_col]
    model = sm.Logit(y, X).fit(disp=False)  # 关闭拟合过程的输出
    return model

def calculate_or_and_ci(model):
    """
    计算 OR 和置信区间
    """
    coef = model.params
    std_err = model.bse
    or_values = np.exp(coef)
    ci_lower = np.exp(coef - 1.96 * std_err)
    ci_upper = np.exp(coef + 1.96 * std_err)
    return pd.DataFrame({
        'OR': or_values,
        '2.5%': ci_lower,
        '97.5%': ci_upper
    })

def split_and_fit_logistic(df, knot, x_col, y_col):
    """
    根据阈值分段并分别拟合逻辑回归模型
    """
    # 分段数据
    df_below = df[df[x_col] < knot]
    df_above = df[df[x_col] >= knot]
    
    # 拟合逻辑回归模型
    model_below = fit_logistic_model(df_below, x_col, y_col)
    model_above = fit_logistic_model(df_above, x_col, y_col)
    
    # 计算 OR 和置信区间
    or_below = calculate_or_and_ci(model_below)
    or_above = calculate_or_and_ci(model_above)
    
    return model_below, model_above, or_below, or_above

# 示例调用
best_knot = 11.4933
model_below, model_above, or_below, or_above = split_and_fit_logistic(df, best_knot, 'cumulative_PHR', 'CVD')

# 打印结果
print("Below Knot Logistic Regression Results:")
print(model_below.summary())
print("\nBelow Knot OR and Confidence Intervals:")
print(or_below)

print("\nAbove Knot Logistic Regression Results:")
print(model_above.summary())
print("\nAbove Knot OR and Confidence Intervals:")
print(or_above)
````

````python
from sklearn.metrics import roc_curve, roc_auc_score
import numpy as np
import matplotlib.pyplot as plt

# 定义 Bootstrap 方法计算 AUC 和置信区间
def bootstrap_auc(y_true, y_score, n_bootstraps=1000, random_seed=42):
    rng = np.random.RandomState(random_seed)
    bootstrapped_scores = []
    
    # 将 y_true 和 y_score 转换为 NumPy 数组
    y_true = np.array(y_true)
    y_score = np.array(y_score)
    
    for i in range(n_bootstraps):
        # 随机采样
        indices = rng.randint(0, len(y_score), len(y_score))
        if len(np.unique(y_true[indices])) < 2:
            # 跳过没有正负样本的情况
            continue
        score = roc_auc_score(y_true[indices], y_score[indices])
        bootstrapped_scores.append(score)
    
    # 计算 95% 置信区间
    sorted_scores = np.sort(bootstrapped_scores)
    ci_lower = sorted_scores[int(0.025 * len(sorted_scores))]
    ci_upper = sorted_scores[int(0.975 * len(sorted_scores))]
    return np.mean(bootstrapped_scores), ci_lower, ci_upper

# 封装绘制 ROC 曲线的函数
def plot_roc_with_ci(df, y_true_col, score_cols, labels, n_bootstraps=1000, random_seed=42):
    """
    绘制多个字段的 ROC 曲线，并计算 AUC 和 95% 置信区间。
    
    参数：
    - df: 数据框
    - y_true_col: 真实标签列名
    - score_cols: 预测分数列名列表
    - labels: 图像中每条曲线的标签列表
    - n_bootstraps: Bootstrap 采样次数
    - random_seed: 随机种子
    """
    y_true = df[y_true_col]
    plt.figure(figsize=(8, 6))
    
    for score_col, label in zip(score_cols, labels):
        y_score = df[score_col]
        
        # 计算 AUC 和置信区间
        auc, ci_lower, ci_upper = bootstrap_auc(y_true, y_score, n_bootstraps, random_seed)

        # 打印 AUC 和置信区间
        print(f"{label} AUC: {auc:.3f} (95% CI: {ci_lower:.3f}–{ci_upper:.3f})")
        
        # 计算 ROC 曲线
        fpr, tpr, _ = roc_curve(y_true, y_score)
        
        # 绘制 ROC 曲线
        plt.plot(fpr, tpr, label=f'{label} (AUC = {auc:.3f}, 95% CI: {ci_lower:.3f}–{ci_upper:.3f})')
    
    # 添加随机猜测的参考线
    plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')
    plt.xlabel('False Positive Rate (FPR)')
    plt.ylabel('True Positive Rate (TPR)')
    plt.title('ROC Curve for CVD Prediction')
    plt.legend(loc='lower right')
    plt.grid()
    plt.show()

# 定义字段名和标签
y_true_col = 'CVD'
score_cols = ['cumulative_PHR', 'PHR_1']  # 需要绘制 ROC 的字段
labels = ['Cumulative PHR', 'Baseline PHR']  # 对应的标签
# 绘制 ROC 曲线
plot_roc_with_ci(df, y_true_col, score_cols, labels)

# 定义字段名和标签
y_true_col = 'CVD'
score_cols = ['plt_1', 'hdl_1', 'PHR_1']  # 需要绘制 ROC 的字段
labels = ['Baseline PLT', 'Baseline HDL-C', 'Baseline PHR']  # 对应的标签
# 绘制 ROC 曲线
plot_roc_with_ci(df, y_true_col, score_cols, labels)
````

### 亚组分析和敏感性分析: 亚组和交互作用

以研究 X 或累积 X 和 CVD 的变化之间的关系是否根据协变量的状态 （性别、婚姻状况、教育水平、吸烟状况、饮酒状况、高血压、糖尿病和血脂异常） 而变化

````python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.formula.api import logit
from scipy import stats  # 新增导入

def subgroup_analysis_with_interaction(df, outcome, predictor, reference_category, subgroup_var):
    """
    分层分析，计算多分类变量的 OR 和 95% CI，并计算交互作用 P 值。
    """
    results = []
    
    # 确保 predictor 类型正确
    df[predictor] = df[predictor].astype(str if isinstance(reference_category, str) else int)
    
    subgroups = df[subgroup_var].unique()
    
    for subgroup in subgroups:
        subgroup_data = df[df[subgroup_var] == subgroup].copy()
        subgroup_data[predictor] = subgroup_data[predictor].astype('category').cat.set_categories(
            [reference_category] + [c for c in subgroup_data[predictor].unique() if c != reference_category],
            ordered=True
        )
        
        formula = f"{outcome} ~ C({predictor}, Treatment({reference_category}))"
        try:
            model = logit(formula, subgroup_data).fit(disp=0)
        except Exception as e:
            print(f"子组 {subgroup} 模型错误: {e}")
            continue
        
        for cat in model.params.index:
            if "T." not in cat:
                continue  # 跳过截距项
            coef = model.params[cat]
            ci = model.conf_int().loc[cat]
            p = model.pvalues[cat]
            
            or_exp = round(np.exp(coef), 2)
            ci_lower = round(np.exp(ci[0]), 2)
            ci_upper = round(np.exp(ci[1]), 2)
            
            results.append({
                "Subgroup": subgroup_var,
                "Category": subgroup,
                "Comparison": cat.split("T.")[-1].strip("]") + f" vs {reference_category}",
                "OR (95% CI)": f"{or_exp} ({ci_lower}–{ci_upper})",
                "P-value": round(p, 3)
            })
    
    # 似然比检验计算交互作用 P 值
    interaction_p = None
    try:
        full_model = logit(f"{outcome} ~ C({predictor}, Treatment({reference_category})) * C({subgroup_var})", df).fit(disp=0)
        reduced_model = logit(f"{outcome} ~ C({predictor}, Treatment({reference_category})) + C({subgroup_var})", df).fit(disp=0)
        lr_stat = 2 * (full_model.llf - reduced_model.llf)
        interaction_p = stats.chi2.sf(lr_stat, full_model.df_model - reduced_model.df_model)
    except Exception as e:
        print(f"交互作用检验失败: {e}")
    
    return pd.DataFrame(results), interaction_p

# 调用函数
results_df, interaction_p = subgroup_analysis_with_interaction(
    df_cleaned, 
    outcome="CVD", 
    predictor="cluster_k3", 
    reference_category=1, 
    subgroup_var="gender_1"
)

# 打印结果
print(results_df)
if interaction_p is not None:
    print(f"Interaction P-value: {interaction_p:.3f}")
else:
    print("Interaction P-value could not be calculated.")
````

````python
def batch_subgroup_analysis(df, outcome, predictor, reference_category, subgroup_vars):
    """
    批量进行分层分析
    
    参数：
    - df: 数据框
    - outcome: 结果变量
    - predictor: 预测变量
    - reference_category: 预测变量的参考类别
    - subgroup_vars: 需要分析的亚组变量列表 (如 ['marital_1', 'hukou_1'])
    
    返回：
    - 包含所有结果的DataFrame
    - 包含交互作用P值的字典
    """

    all_results = []
    interaction_p_dict = {}
    
    for subgroup_var in subgroup_vars:
        # 新增列名检查
        if predictor not in df.columns:
            raise ValueError(f"预测变量 {predictor} 不存在于数据框中")
        if subgroup_var not in df.columns:
            raise ValueError(f"分层变量 {subgroup_var} 不存在于数据框中")
        if outcome not in df.columns:
            raise ValueError(f"结果变量 {outcome} 不存在于数据框中")
        try:
            # 执行单次分析
            results_df, interaction_p = subgroup_analysis_with_interaction(
                df, 
                outcome=outcome,
                predictor=predictor,
                reference_category=reference_category,
                subgroup_var=subgroup_var
            )
            
            # 合并结果
            if not results_df.empty:
                all_results.append(results_df)
                interaction_p_dict[subgroup_var] = interaction_p
        except Exception as e:
            print(f"亚组分析失败: {subgroup_var}，错误信息: {str(e)}")
            continue
    
    # 合并所有结果
    final_results = pd.concat(all_results, ignore_index=True) if all_results else pd.DataFrame()
    return final_results, interaction_p_dict

# 使用示例
subgroup_vars = ['gender_1', 'marital_1', 'hukou_1', 'edu_1', 'smoking_status_1', 'drinking_status_1', 
                 'hypertension_1',  # 高血压
                 'diabetes_1',  # 糖尿病
                 'dyslipidemia_1' # 血脂异常
                 ]

# 执行批量分析
all_results, interaction_ps = batch_subgroup_analysis(
    df_cleaned,
    outcome="CVD",
    predictor="cluster_k3",
    reference_category=1,
    subgroup_vars=subgroup_vars
)

# 打印结果
print("所有亚组分析结果:")
print(all_results)

print("\n交互作用P值:")
for var, p in interaction_ps.items():
    print(f"{var}: {p:.3f}" if p is not None else f"{var}: 无法计算")
````

````python
health1_path = r"C:\Users\aa\Downloads\charls_data\health_status_and_functioning1\health_status_and_functioning.dta"
columns={
    'ID': 'id',
    'da007_7_': '1_heart', # 1:yes, 2:no
    'da009_7_': '1_heart_time_type', # 1:year, 2:age
    'da009_1_7_': '1_heart_year',
    'da009_2_7_': '1_heart_age',
    'da007_8_': '1_stroke', # 1:yes, 2:no
    'da009_8_': '1_stroke_time_type', # 1:year, 2:age
    'da009_1_8_': '1_stroke_year',
    'da009_2_8_': '1_stroke_age',
    'da007_2_': '1_dyslipidemia', # 血脂异常 1:yes, 2:no
    'da009_2_': '1_dyslipidemia_time_type', # 1:year, 2:age
    'da009_1_2_': '1_dyslipidemia_year',
    'da009_2_2_': '1_dyslipidemia_age',
    'da007_1_': '1_hypertension', #高血压 1:yes, 2:no  
    'da009_1_': '1_hypertension_time_type', # 1:year, 2:age
    'da009_1_1_': '1_hypertension_year',
    'da009_2_1_': '1_hypertension_age',
    'da007_3_': '1_diabetes', #糖尿病 1:yes, 2:no
    'da007_4_': '1_cancer', #癌症 1:yes, 2:no
    'da059': '1_smoke', # 1:吸烟, 2:不吸烟
    'da061': '1_smoking', # 1:仍在吸烟, 2:已戒烟, None:不吸烟
    'da069': '1_drink', # 1:从不, 2\3: others
}
health1_df = get_data(health1_path, columns)

columns_to_apply = ['1_heart', '1_stroke', '1_dyslipidemia', '1_hypertension', '1_diabetes', '1_cancer']  # 需要应用函数的列名
for col in columns_to_apply:
    health1_df[col] = health1_df[col].apply(one_otherTo1_0)
# 新增字段 1_cvd = 1_heart or 1_stroke
health1_df['1_cvd'] = health1_df.apply(
    lambda row: 1 if (row['1_heart'] == 1 or row['1_stroke'] == 1) else 0, axis=1
)
# 计算 高血压(hypertension) 的比例
hypertension_counts_1 = health1_df['1_hypertension'].value_counts(normalize=True)
print(hypertension_counts_1)
# 计算 血脂异常(dyslipidemia) 的比例
dyslipidemia_counts_1 = health1_df['1_dyslipidemia'].value_counts(normalize=True)
print(dyslipidemia_counts_1)

# 将吸烟状态分为三类： 1_smoking_status: 从未吸烟、以前吸烟、当前吸烟
# 定义分类函数
def classify_smoking(row):
    if row['1_smoke'] == 2:
        return 'Never'
    elif row['1_smoke'] == 1 and row['1_smoking'] == 2:
        return 'Former'
    elif row['1_smoke'] == 1 and row['1_smoking'] == 1:
        return 'Current'
    # else: 不处理就是相当于 None
    #     return 'Unknown'  # 处理缺失值或异常情况
health1_df['1_smoking_status'] = health1_df.apply(classify_smoking, axis=1)
# 计算 1_smoking_status 的比例
smoking_status_1 = health1_df['1_smoking_status'].value_counts(normalize=True)
print(smoking_status_1)

health1_df['1_drinking_status'] = health1_df['1_drink'].apply(lambda x: 'Others' if x == 1 else 'Current') # 1:从不, 2\3: others -> Current, Others
# print(health1_df.info())
print(health1_df)
````

````python
# 去掉存在缺失值的行, 不考虑 1_heart_year, 1_heart_age, 1_stroke_year, 1_stroke_age, 1_dyslipidemia_year, 1_dyslipidemia_age, 1_hypertension_year, 1_hypertension_age, 3_heart_year, 3_heart_age, 3_stroke_year, 3_stroke_age, 3_dyslipidemia_year, 3_dyslipidemia_age, 3_hypertension_year, 3_hypertension_age 字段

# 选择要研究的字段 
fields_to_keep = [
    'id', 
    '1_age',  '1_gender', # 模型1
    # '3_age',
    '1_marital', '1_hukou', '1_edu', '1_smoking_status', '1_drinking_status', # 模型2(另外增加的)
    # '3_smoking_status', '3_drinking_status', 
    '1_hypertension', '1_diabetes', '1_dyslipidemia', '1_cancer', # 模型3(另外增加的)
    # '3_hypertension', '3_diabetes', '3_dyslipidemia', '3_cancer',
    '1_SBP', '1_DBP', '1_tc', '1_hdl', '1_ldl', '1_hba1c', # 模型4(另外增加的)
    # '3_SBP', '3_DBP', '3_tc', '3_hdl', '3_ldl', '3_hba1c',
    '1_glu', '1_tg', '1_h', '1_waist',
    '1_cvd',
    '3_glu', '3_tg', '3_h', '3_waist',
    '3_cvd', '4_cvd', '5_cvd', # 模型5(另外增加的)
    '1_weight', '3_weight', # 其他常用变量
]
df = pd.read_csv("./charls.csv", usecols=fields_to_keep)
# 1. 去掉存在缺失值的行
# 告诉我哪些行存在缺失
print(df.isnull().sum())
df = df.dropna()
print(df.shape)
# 2. 去除 1_cvd = 1 的行
df = df[df['1_cvd'] == 0]
# cvd = 3_cvd or 4_cvd or 5_cvd
df['CVD'] = df['3_cvd'] | df['4_cvd'] | df['5_cvd']
# 计算 cvd 的比例
cvd_counts_3 = df['3_cvd'].value_counts(normalize=True)
print(cvd_counts_3)
cvd_counts_4 = df['4_cvd'].value_counts(normalize=True)
print(cvd_counts_4)
cvd_counts_5 = df['5_cvd'].value_counts(normalize=True)
print(cvd_counts_5)
cvd_counts = df['CVD'].value_counts(normalize=True)
print(cvd_counts)
# 3. 去除 身高, 腰围 异常的行, 例如过大或过小, 可能是测量错误
# 假设身高 (1_h, 3_h) 的合理范围是 100cm 到 250cm
# 假设腰围 (1_waist, 3_waist) 的合理范围是 40cm 到 150cm
df = df[
    (df['1_h'] >= 100) & (df['1_h'] <= 250) &  # 身高范围
    (df['3_h'] >= 100) & (df['3_h'] <= 250) &  # 身高范围
    (df['1_waist'] >= 40) & (df['1_waist'] <= 150) &  # 腰围范围
    (df['3_waist'] >= 40) & (df['3_waist'] <= 150)    # 腰围范围
]
print(f"After removing rows with abnormal height or waist: {df.shape}")
print(df)
import numpy as np
# 1_TyG = ln [1_tg (mg/dl) ×  1_glu (mg/dl)/2], WHtR = 腰围/身高, TyG_WHtR = TyG × WHtR
# 计算 TyG, WHtR, TyG_WHtR
df['1_TyG'] = np.log(df['1_tg'] * df['1_glu'] / 2)
df['1_WHtR'] = df['1_waist'] / df['1_h']
df['1_TyG_WHtR'] = df['1_TyG'] * df['1_WHtR']
# 计算 波次3 的 TyG, WHtR, TyG_WHtR
df['3_TyG'] = np.log(df['3_tg'] * df['3_glu'] / 2)
df['3_WHtR'] = df['3_waist'] / df['3_h']
df['3_TyG_WHtR'] = df['3_TyG'] * df['3_WHtR']
# cumulative TyG_WHtR = (1_TyG_WHtR + 3_TyG_WHtR)/2 * (2015-2012+1)
df['cumulative_TyG_WHtR'] = (df['1_TyG_WHtR'] + df['3_TyG_WHtR']) / 2 * (2015 - 2012 + 1)

# 保存结果到 CSV 文件
df.to_csv("charls_TyG_WHtR.csv", index=False, encoding='utf-8-sig')
````

### 限制三次样条函数 （RCS） 回归模型来探索累积 TyG-WHtR 和 CVD 事件的潜在非线性关联

````python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from patsy import dmatrix, cr # pip install patsy
import matplotlib.pyplot as plt

# 选择要研究的字段 
fields_to_keep = [
    # 'id', 
    'cumulative_TyG_WHtR','cvd'
]

df = df_encoded[fields_to_keep]

# 手动指定节点位置（关键！确保训练和预测一致）
knots = np.percentile(df['cumulative_TyG_WHtR'], [25, 50, 75])  # 3个节点，生成2个基函数（不含截距）
spline_basis = dmatrix(
    cr(df['cumulative_TyG_WHtR'], knots=knots, constraints='center'),
    data=df,
    return_type='dataframe'
)
# 检查生成的列数（应为4列）
print("样条基矩阵列数:", spline_basis.shape[1])  # 输出应为4
# 重命名列（根据实际列数，例如4列）
# print(spline_basis.columns)  # 查看列名
spline_basis.columns = [f'spline{i}' for i in range(spline_basis.shape[1])]  # spline0, spline1, spline2, spline3
# 合并基函数到原始数据
df_spline = pd.concat([df, spline_basis], axis=1)

# 提取样条基列名（假设生成列名为 'cr(x, df=4)[x]' 等）
spline_cols = [col for col in spline_basis.columns if col != 'spline0']
print("样条基列名:", spline_cols)  # 输出应为 ['spline0', 'spline1', 'spline2', 'spline3']
# 构建模型公式
formula = 'cvd ~ ' + ' + '.join(spline_cols)
# formula = 'cvd ~ spline1 + spline2'  # 根据实际列数调整

# 拟合模型
model = sm.GLM.from_formula(
    formula=formula,
    data=df_spline,
    family=sm.families.Binomial()
).fit()

# 输出模型摘要
print(model.summary())
````

````python
# 生成均匀分布的自变量值
x_pred = np.linspace(
    df['cumulative_TyG_WHtR'].min(),
    df['cumulative_TyG_WHtR'].max(),
    300
)

# 生成预测用的样条基矩阵（使用相同节点）
spline_pred = dmatrix(
    cr(x_pred, knots=knots, constraints='center'),
    {'cumulative_TyG_WHtR': x_pred},
    return_type='dataframe'
)
spline_pred.columns = [f'spline{i}' for i in range(spline_pred.shape[1])]  # 列名必须与训练一致

# 构建预测数据框（仅保留模型所需的列）
df_pred = spline_pred[spline_cols]  # 仅包含 spline1, spline2, spline3
# 获取预测结果的置信区间（默认95%）
pred_results = model.get_prediction(df_pred)
pred_ci = pred_results.conf_int()  # 返回 [lower, upper]

# 提取预测值和置信区间上下限
pred = pred_results.predicted_mean
lower = pred_ci[:, 0]
upper = pred_ci[:, 1]

# 绘图代码（修正后）
plt.figure(figsize=(10, 6))
sns.lineplot(x=x_pred, y=pred, color='blue', lw=2)
plt.fill_between(
    x_pred,
    lower,  # 使用置信区间下限
    upper,  # 使用置信区间上限
    alpha=0.2,
    color='blue'
)
plt.xlabel('cumulative_TyG_WHtR')
plt.ylabel('Probability of CVD')
plt.title('RCS Association Plot')
plt.show()
````

````python
# 生成预测数据（同上）
x_pred = np.linspace(
    df['cumulative_TyG_WHtR'].min(),
    df['cumulative_TyG_WHtR'].max(),
    300
)
spline_pred = dmatrix(
    cr(x_pred, knots=knots, constraints='center'),
    {'cumulative_TyG_WHtR': x_pred},
    return_type='dataframe'
)
spline_pred.columns = [f'spline{i}' for i in range(spline_pred.shape[1])]
df_pred = spline_pred[spline_cols]

# 获取线性预测值（Log Odds）
log_odds = model.predict(df_pred, linear=True)  # linear=True 返回 η = Xβ

# 手动计算标准误（关键步骤！）
# 1. 获取模型参数协方差矩阵
vcov = model.cov_params()
# 2. 提取预测数据的设计矩阵（X）
X_pred = model.model.exog[:df_pred.shape[0], :]  # 确保与 df_pred 对应
# 3. 计算每个预测点的方差：Var(η) = X * Vcov * X.T
var_log_odds = np.diag(X_pred @ vcov @ X_pred.T)
# 4. 计算标准误
se_log_odds = np.sqrt(var_log_odds)

# 计算置信区间
lower_log_odds = log_odds - 1.96 * se_log_odds
upper_log_odds = log_odds + 1.96 * se_log_odds

# 绘制 Log Odds 曲线
plt.figure(figsize=(10, 6))
sns.lineplot(x=x_pred, y=log_odds, color='blue', lw=2)
plt.fill_between(
    x_pred,
    lower_log_odds,
    upper_log_odds,
    alpha=0.2,
    color='blue'
)
plt.xlabel('cumulative_TyG_WHtR')
plt.ylabel('Log Odds of CVD')
plt.title('RCS Association Plot (Log Odds)')
plt.show()
````

### ROC 曲线分析

The ROC curve indicated that cumulative TyG-WHtR (AUC 0.596, 95%CI 0.572–0.621) had the highest diagnostic efficacy for CVD, followed by TyG-WHtR at baseline (AUC 0.596, 95%CI 0.572–0.621), WHtR at baseline (AUC 0.584, 95%CI 0.559–0.609), and TyG at baseline (AUC 0.568, 95%CI 0.544–0.593).(Fig. 4) Cumulative TyG-WHtR also had the highest diagnostic efficacy for heart disease and stroke. (Additional file 1: Figures S2, S3).
使用受试者工作特征曲线（ROC）分析来评估 TyG-WHtR 在预测 CVD 方面的有效性。ROC 曲线下面积（AUC）用于量化模型的预测能力。AUC 值范围从 0 到 1，值越接近 1 表示模型的预测能力越强。我们还计算了不同 TyG-WHtR 分类的 AUC 值，以比较其在 CVD 风险评估中的有效性。

````python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# 示例数据
np.random.seed(42)

# 定义指标
features = ['1_TyG', '1_WHtR', '1_TyG_WHtR', 'cumulative_TyG_WHtR']

# 绘制 ROC 曲线
plt.figure(figsize=(8, 6))
for feature in features:
    # 计算 ROC 曲线
    fpr, tpr, _ = roc_curve(df['CVD'], df[feature])
    auc = roc_auc_score(df['CVD'], df[feature])
    
    # 绘制曲线
    plt.plot(fpr, tpr, label=f'{feature} (AUC = {auc:.3f})')

# 绘制对角线
plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')

# 图形美化
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt.title('ROC Curve for Predicting CVD')
plt.legend(loc='lower right')
plt.grid()
plt.show()
````

````python
from sklearn.metrics import roc_curve, auc, roc_auc_score
from sklearn.utils import resample

def calculate_auc_ci(y_true, y_score, n_bootstraps=1000, ci=95):
    """计算AUC及其置信区间（自助法）"""
    auc_scores = []
    rng = np.random.RandomState(42)
    
    # 自助法重采样
    for _ in range(n_bootstraps):
        indices = resample(np.arange(len(y_true)), random_state=rng)
        if len(np.unique(y_true[indices])) < 2:
            continue  # 跳过无效样本
        auc_score = roc_auc_score(y_true[indices], y_score[indices])
        auc_scores.append(auc_score)
    
    # 计算置信区间
    lower = np.percentile(auc_scores, (100 - ci) / 2)
    upper = np.percentile(auc_scores, ci + (100 - ci) / 2)
    mean_auc = np.mean(auc_scores)
    
    return mean_auc, (lower, upper)

# 计算各指标的AUC和CI
metrics = ['cumulative_TyG_WHtR', '1_TyG_WHtR', '1_WHtR', '1_TyG']
results = {}

for metric in metrics:
    y_true = df['cvd']
    y_score = df[metric]
    auc_mean, auc_ci = calculate_auc_ci(y_true, y_score)
    results[metric] = {
        'AUC': auc_mean,
        'CI_lower': auc_ci[0],
        'CI_upper': auc_ci[1]
    }

# 打印结果
for metric, res in results.items():
    print(f"{metric}: AUC = {res['AUC']:.3f} (95% CI {res['CI_lower']:.3f}-{res['CI_upper']:.3f})")
````
